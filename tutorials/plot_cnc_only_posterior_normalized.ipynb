# CNC-only posterior constraints (normalized)

Same plotting style as `plot_combine_posterior_normalized.ipynb`, but only the CNC contour.


```python
import numpy as np
from getdist.mcsamples import loadMCSamples
from getdist import plots
import matplotlib.pyplot as plt
```

```python
chain_root_cnc = "/scratch/scratch-lxu/compute_packages/cosmocnc_jax/tutorials/chains/cnc_planck_scatter_jax"
samples_cnc = loadMCSamples(chain_root_cnc, settings={"ignore_rows": 0.3})
print(f"Loaded CNC samples: {samples_cnc.numrows}")

fid = {
    'Omega_m':     0.309576,
    'H0':          67.66,
    'ln10_10A_s':  np.log(1.9527109742006746e-09 * 1e10),
    'n_s':         0.9665,
    'omega_b':     0.02242,
    'omega_cdm':   0.1193,
    'A_SZ':        -4.23820656,
    'alpha_SZ':    1.12,
    'sigma_lnY':   0.173,
}
```

```python
params = ['Omega_m', 'H0', 'ln10_10A_s', 'n_s', 'omega_b', 'omega_cdm', 'A_SZ', 'alpha_SZ', 'sigma_lnY']
stats = samples_cnc.getMargeStats()
for p in params:
    lim = stats.parWithName(p)
    if lim is not None:
        print(f"{p:12s}: mean={lim.mean:.6g}, sigma={lim.err:.6g}")
```

```python
plot_params = ['Omega_m', 'H0', 'ln10_10A_s', 'A_SZ', 'alpha_SZ', 'sigma_lnY']

g = plots.get_subplot_plotter(width_inch=12)
g.settings.figure_legend_frame = False
g.settings.alpha_filled_add = 0.70
g.settings.axes_fontsize = 12
g.settings.lab_fontsize = 13
g.settings.legend_fontsize = 13

g.triangle_plot(
    [samples_cnc],
    plot_params,
    filled=True,
    contour_colors=['#1f77b4'],
    legend_labels=['CNC only'],
    markers={k: fid[k] for k in plot_params if k in fid},
)

plt.tight_layout()
plt.savefig('/scratch/scratch-lxu/compute_packages/cosmocnc_jax/tutorials/triangle_cnc_only.png', dpi=300)
plt.show()
```
